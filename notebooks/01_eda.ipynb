# EDA —  Credit Card Fraud Dataset


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/creditcard.csv')
df.shape
```

```python
# Class imbalance
counts = df['Class'].value_counts()
pct = df['Class'].value_counts(normalize=True) * 100
print(counts)
print(pct)

fig, ax = plt.subplots(figsize=(4,4))
counts.plot(kind='bar', ax=ax)
ax.set_yscale('log')
ax.set_xticklabels(['Legit', 'Fraud'], rotation=0)
ax.set_title('Class distribution (log scale)')
plt.show()
```

```python
# Amount distribution by class
fig, ax = plt.subplots(1, 2, figsize=(10,4))
sns.histplot(df[df.Class==0]['Amount'], bins=50, ax=ax[0]).set_title('Legit — Amount')
sns.histplot(df[df.Class==1]['Amount'], bins=50, ax=ax[1], color='red').set_title('Fraud — Amount')
plt.tight_layout()
plt.show()
```

```python
# Correlation of PCA components with Class
corr = df.corr(numeric_only=True)['Class'].drop('Class').sort_values()
fig, ax = plt.subplots(figsize=(6,8))
corr.plot(kind='barh', ax=ax)
ax.set_title('Feature correlation with fraud label')
plt.tight_layout()
plt.show()
```
